```python
file_path = "/kaggle/input/datasets/shree0910/ai-and-data-science-job-market-dataset-20202026/AI Job Market Dataset.csv"
```

```python
df = pd.read_csv(file_path)
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
df = df.dropna()
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

```

```python
print("\nAverage Salary:", df['salary'].mean())
print("Highest Salary:", df['salary'].max())
print("Lowest Salary:", df['salary'].min())
```

```python
print("\nTop Job Titles:")
print(df['job_title'].value_counts().head())
```

```python
print("\nTop Countries:")
print(df['country'].value_counts().head())
```

```python
skills = ['skills_python','skills_sql','skills_ml','skills_deep_learning','skills_cloud']
print("\nSkills Demand:")
print(df[skills].sum())
```

```python
print("\nAverage Experience:", df['years_experience'].mean())
```
